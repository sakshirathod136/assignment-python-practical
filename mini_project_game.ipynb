```python

import random
def play_best_of_five():
    print("========================================")
    print(" Welcome to Rock, Paper, Scissors!       ")
    print(" Match Format: First to 3 wins (Best of 5) ")
    print("=========================================\n")
    choices = ["rock", "paper", "scissors"]
    # Initialize the scores
    player_score = 0
    computer_score = 0
    round_number = 1
    # The loop runs as long as BOTH scores are under 3
    while player_score < 3 and computer_score < 3:
        print(f"--- Round {round_number} ---")
        user_choice = input("Enter rock, paper, or scissors: ").lower()
        # Validate input
        if user_choice not in choices:
            print("❌ Invalid choice! Please pick rock, paper, or scissors.\n")
            continue
        computer_choice = random.choice(choices)
        print(f"You chose: {user_choice}")
        print(f"Computer chose: {computer_choice}")
        # Determine round winner
        if user_choice == computer_choice:
            print("👉 It's a tie for this round!")
        elif (user_choice == "rock" and computer_choice == "scissors") or \
             (user_choice == "paper" and computer_choice == "rock") or \
             (user_choice == "scissors" and computer_choice == "paper"):
            print("🥇 You won this round!")
            player_score += 1
        else:
            print("🤖 Computer won this round!")
            computer_score += 1
        # Display the current scorecard
        print(f"📊 SCOREBOARD: You: {player_score} | Computer: {computer_score}")
        print("-" * 30 + "\n")
        round_number += 1
    # The loop broke, meaning someone reached 3 wins. Final match results:
    print("=========================================")
    print("               GAME OVER                 ")
    print("=========================================")
    if player_score == 3:
        print(f"🎉 Amazing! You won the match {player_score} to {computer_score}!")
    else:
        print(f"😢 The computer won the match {computer_score} to {player_score}. Better luck next time!")
    print("=========================================")
# Run the tournament
play_best_of_five()







```
